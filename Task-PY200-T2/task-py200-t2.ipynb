```python
try:
    number = float(input("Enter a number: "))

    if number > 0:
        print(f"Your entered number {number} is positive.")
    elif number < 0 :
        positive_number = abs(number)
        print(f"Your entered number {number} was negative, however positive value of your number is {positive_number}")
    else:
        print(f"The number is Zero!")

except ValueError:
    print("Invalid input! Please enter a numeric value.")
```
