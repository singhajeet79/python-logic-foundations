```python
# Task PY200_T4: Password Strength Checker
s = "pwdMon1!"  # Change this to test different passwords
```

```python
# 1. Length Check
length_ok = len(s) >= 8
print(f"{length_ok} {len(s)}")
```

```python
# 2, 3, 4. Character Content Checks
has_digit = False
has_upper = False
has_lower = False

for char in s:
    if char.isdigit():
        has_digit = True
    if char.isupper():
        has_upper = True
    if char.islower():
        has_lower = True
```

```python
# --- Final Validation ---
if length_ok and has_digit and has_upper and has_lower:
    print("STRONG")
else:
    print("WEAK")
```
