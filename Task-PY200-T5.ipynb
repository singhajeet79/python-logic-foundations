# Chessboard with Numbers Input: N Print an N x N grid
**Input:** N (integer)

**Rules:** Diagonal is 'X', (row+col) even is '1', else '0'.

```python
N = int(input("N: "))

for r in range(N):
    row_string = ""
    for c in range(N):
        if r == c:
            char = "X"
        elif (r + c) % 2 == 0:
            char = "1"
        else:
            char = "0"
        
        row_string += char + " "
    print(row_string)
```
