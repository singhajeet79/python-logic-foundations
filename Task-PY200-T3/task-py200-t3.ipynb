```python
# Task PY200_T3: Electricity Bill Slabs
units = int(input("Enter units: "))
total_bill = 0
```

```python
remaining_units = units
# Slab 1: First 100 units @ Rs. 2
if remaining_units > 100:
    total_bill = total_bill + 100 * 2
    remaining_units = remaining_units - 100
else:
    total_bill = total_bill + remaining_units * 2
    remaining_units = 0
```

```python
# Slab 2: Next 100 (100-200) units @ 3
if remaining_units > 100:
    total_bill = total_bill + 100 * 3
    remaining_units = remaining_units - 100
else:
    total_bill = total_bill + remaining_units * 3
    remaining_units = 0
```

```python
# Slab 3: Next 300 (200-500) units @ Rs 5
if remaining_units > 300:
    total_bill = total_bill + 300 * 5
    remaining_units = remaining_units - 300
else:
    total_bill = total_bill + remaining_units * 5
    remaining_units = 0
```

```python
# Slab 4: Above 500 @ Rs. 8
if remaining_units > 0:
    total_bill = total_bill + remaining_units * 8

# Add fixed charge if any units were consumed
if units > 0:
    total_bill = total_bill + 50
```

```python
# Print Output ---
print(f"Total Units: {units}")
print(f"Total Bill:  {total_bill}")
```
