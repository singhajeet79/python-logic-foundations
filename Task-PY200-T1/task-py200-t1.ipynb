# Mean, Median, Mode using simple maths

```python
data = [12, 10, 16, 10, 19, 14, 22, 10]
n = len(data)
```

```python
# --- 1. MEAN (The Average) ---
# Sum all numbers and divide by the count
total_sum = 0
for x in data:
    total_sum += x
mean = total_sum / n
print(mean)
```

```python
n = len(data)
for i in range(n):
    for j in range(0, n-i-1):
        if data[j] > data[j+1]:
            # Swap elements if they are not in the ascending order
            temp = data[j]
            data[j] = data[j+1]
            data[j+1] = temp
if n % 2 == 0:
    median = (data[n//2 - 1] + data[n//2]) / 2
else:
    median = data[n//2]
print(median)
    
```

```python
counts = {}
for x in data:
    if x in counts:
        counts[x] += 1
    else:
        counts[x] = 1
max_count = 0
mode = None
for i in counts:
    if counts[i] > max_count:
        max_count = counts[i]
        mode = i
print(mode)
```

```python
# --- Final Output ---
print(f"Mean:   {mean}\n")
print(f"Median: {median}\n")
print(f"Mode:   {mode}")
```
